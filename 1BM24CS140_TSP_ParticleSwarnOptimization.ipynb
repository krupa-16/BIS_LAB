```python

import random

# Distance matrix
dist = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 17],
    [15, 35, 0, 30, 28],
    [20, 25, 30, 0, 12],
    [25, 17, 28, 12, 0]
]

n = len(dist)
particles = 10
iterations = 100

# Calculate total route distance
def fitness(route):
    total = 0
    for i in range(n):
        total += dist[route[i]][route[(i + 1) % n]]
    return total

# Generate random route
def create_route():
    route = list(range(n))
    random.shuffle(route)
    return route

# Swap two cities
def swap(route):
    a, b = random.sample(range(n), 2)
    route[a], route[b] = route[b], route[a]
    return route

# Initialize particles
positions = [create_route() for _ in range(particles)]
pbest = [route[:] for route in positions]
pbest_cost = [fitness(route) for route in positions]

# Initialize global best
best_index = pbest_cost.index(min(pbest_cost))
gbest = pbest[best_index][:]
gbest_cost = pbest_cost[best_index]

# PSO iterations
for _ in range(iterations):
    for i in range(particles):

        # Move towards personal best
        if random.random() < 0.5:
            positions[i] = swap(positions[i])

        # Move towards global best
        if random.random() < 0.5:
            positions[i] = gbest[:]

        # Calculate new fitness
        cost = fitness(positions[i])

        # Update personal best
        if cost < pbest_cost[i]:
            pbest[i] = positions[i][:]
            pbest_cost[i] = cost

        # Update global best
        if cost < gbest_cost:
            gbest = positions[i][:]
            gbest_cost = cost

print("Best Route:", [city + 1 for city in gbest] + [gbest[0] + 1])
print("Minimum Distance:", gbest_cost)
```
