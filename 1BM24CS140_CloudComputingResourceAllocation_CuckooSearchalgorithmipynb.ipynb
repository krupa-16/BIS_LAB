```python

import random

# Cloud tasks and their processing times
tasks = [4, 6, 3, 5, 2]

# Number of virtual machines
vm_count = 3

# CSA parameters
nests_count = 15
iterations = 100
pa = 0.25

# Calculate makespan
def fitness(solution):
    loads = [0] * vm_count

    for i in range(len(tasks)):
        loads[solution[i]] += tasks[i]

    return max(loads)

# Generate random allocation
def create_nest():
    return [random.randint(0, vm_count - 1)
            for _ in tasks]

# Generate new solution using random flight
def levy_flight(solution):
    new = solution[:]

    step = abs(random.gauss(0, 1))
    changes = max(1, min(len(tasks), int(step * len(tasks))))

    for _ in range(changes):
        index = random.randint(0, len(tasks) - 1)
        new[index] = random.randint(0, vm_count - 1)

    return new

# Main Cuckoo Search Algorithm
random.seed(42)

nests = [create_nest() for _ in range(nests_count)]

best = min(nests, key=fitness)[:]

for _ in range(iterations):

    # Generate new solutions
    for i in range(nests_count):
        new = levy_flight(nests[i])

        j = random.randint(0, nests_count - 1)

        if fitness(new) < fitness(nests[j]):
            nests[j] = new

        if fitness(new) < fitness(best):
            best = new[:]

    # Abandon worst nests
    count = max(1, int(pa * nests_count))

    worst = sorted(
        range(nests_count),
        key=lambda i: fitness(nests[i]),
        reverse=True
    )[:count]

    for i in worst:
        nests[i] = create_nest()

    # Update global best
    current = min(nests, key=fitness)

    if fitness(current) < fitness(best):
        best = current[:]

# Display allocation
loads = [0] * vm_count

print("Optimal Resource Allocation:")

for i in range(len(tasks)):
    vm = best[i]
    loads[vm] += tasks[i]
    print("Task T" + str(i+1),
          "-> VM" + str(vm+1))

print("VM Loads:", loads)
print("Minimum Makespan:", fitness(best))
```
