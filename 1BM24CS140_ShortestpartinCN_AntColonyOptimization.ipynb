```python

import random

# Network topology (routers and link costs)
graph = {
    0: {1: 2, 2: 4},
    1: {0: 2, 2: 1, 3: 7},
    2: {0: 4, 1: 1, 3: 3, 4: 5},
    3: {1: 7, 2: 3, 4: 2},
    4: {2: 5, 3: 2}
}

source = 0
destination = 4
ants = 10
iterations = 50
alpha = 1
beta = 2
rho = 0.5
Q = 100

# Initialize pheromone
pheromone = {}
for u in graph:
    for v in graph[u]:
        pheromone[(u, v)] = 1.0

best_path = None
best_cost = float('inf')

# Calculate path cost
def path_cost(path):
    return sum(graph[path[i]][path[i+1]]
               for i in range(len(path)-1))

# Construct ant path
def create_path():
    path = [source]
    current = source

    while current != destination:
        neighbors = [
            v for v in graph[current] if v not in path
        ]

        if not neighbors:
            return None

        weights = []
        for v in neighbors:
            p = pheromone[(current, v)] ** alpha
            h = (1 / graph[current][v]) ** beta
            weights.append(p * h)

        next_node = random.choices(
            neighbors, weights=weights, k=1
        )[0]

        path.append(next_node)
        current = next_node

    return path

random.seed(42)

for _ in range(iterations):
    paths = []

    # Ants construct paths
    for _ in range(ants):
        path = create_path()

        if path is not None:
            cost = path_cost(path)
            paths.append((path, cost))

            if cost < best_cost:
                best_cost = cost
                best_path = path[:]

    # Pheromone evaporation
    for edge in pheromone:
        pheromone[edge] *= (1 - rho)

    # Pheromone deposition
    for path, cost in paths:
        deposit = Q / cost

        for i in range(len(path)-1):
            edge = (path[i], path[i+1])
            pheromone[edge] += deposit

# Display result
print("Shortest Path:",
      " -> ".join("R" + str(x+1) for x in best_path))
print("Minimum Cost:", best_cost)
```
